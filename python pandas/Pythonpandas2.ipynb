```python
#10.concatenate and merge Dataframe(joins)
```

```python
df1=pd.DataFrame({'ID':[1,2,3,4],'Name':['A','B','C','D']})
df1
```

```python
df2=pd.DataFrame({'ID':[1,2,2,4],'Score':[88,89,90,91]})
df2
```

```python
pd.concat([df1,df2],axis=0)  #vertical or row level join
```

```python
pd.concat([df1,df2],axis=1)   #column level join/horizontal
```

```python
#merge=join
pd.merge(df1,df2,how='inner',on='ID')
```

```python
pd.merge(df1,df2,how='left',on='ID')
```

```python
pd.merge(df1,df2,how='inner', left_on='ID',right_on='ID' )
```
