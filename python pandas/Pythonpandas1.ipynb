```python
#6.rows and column-operations(add,update,delete)
data = {
    'Name':['Bishwas','Vishika','lalita','Rohit'],
    'Age':[16,11,22,33],
    'salary':[30000,300000,400000,50000]

}
import pandas as pd
```

```python
df= pd.DataFrame(data)
print(df)
```

```python
#add new column
df['team']=['CEO','HR','CTO','DA']
df
```

```python
df['bonus']= df['salary']*0.2
df
```

```python
#add new row
df.loc[len(df)] = ['ABC',21,30000,'BA',40000]
df
```

```python
#update value
df.loc[0,'salary']=50000
df
```

```python
df.loc[df.Name =='Bishwas','salary']=40000
df
```

```python
#delete value
df.drop('bonus',axis=1,inplace=True)   #drop cloumn
df
```

```python
df=df.drop(df[df.Name == 'ABC'].index) #drop row
df
```

```python
#df.drop(['Bonus','Team'],axis=1)   to delete multiple column
```

```python
#sort values
df.sort_values('salary')  #ascending order
```

```python
df.sort_values('salary', ascending=False)  #descending order
```
